```python
!pip install accelerate -U
!pip install transformers
!pip install ai-energymeter

import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    GenerationConfig,
    TextGenerationPipeline
)
from energymeter import EnergyMeter
from IPython.display import Markdown


#!sudo chmod -R a+r /sys/class/powercap/intel-rapl
```

# Running models

```python
# Load model and tokenizer once
#model_id = "TucanoBR/Tucano-1b1-Instruct"
#model_id = "Polygl0t/Tucano2-qwen-0.5B-Instruct"
#model_id = "Polygl0t/Tucano2-qwen-1.5B-Instruct"
#model_id = "Polygl0t/Tucano2-qwen-3.7B-Instruct"
model_id = "nicholasKluge/TeenyTinyLlama-460m-Chat"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

generation_config = GenerationConfig(
    do_sample=True,
    max_new_tokens=1024,
    renormalize_logits=True,
    repetition_penalty=1.2,
    temperature=0.7,
    top_k=100,
    top_p=1.0,
    use_cache=True,
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

generator = TextGenerationPipeline(
    model=model,
    task="text-generation",
    tokenizer=tokenizer,
    device=device
)

prompt = """
Responda à pergunta abaixo de forma direta e objetiva.

Pergunta: Qual cidade é a capital do estado do Rio Grande do Sul?

Responda apenas com o nome da cidade.
"""

# Run 10 iterations
num_iterations = 10

gpu_energy = []
cpu_energy = []
disk_energy = []
dram_energy = []
total_energy = []

for i in range(num_iterations):

    print(f"\n===== Iteration {i+1}/{num_iterations} =====")

    em = EnergyMeter(
        disk_avg_speed=2700 * 1e6,
        disk_active_power=1.98,
        disk_idle_power=1.42,
        label=f"PCH-ALDPRO-512-iteration-{i+1}",
        include_idle=True
    )

    # Start measurement
    em.begin()

    # Inference
    completions = generator(
        f"<instruction>{prompt}</instruction>",
        generation_config=generation_config
    )

    # Stop measurement
    em.end()

    # Print completions
    print(completions[0]['generated_text'])
    for comp in completions:
      display(Markdown(comp["generated_text"]))

    # Get energy values
    cpu = em.get_total_joules_cpu()
    disk = em.get_total_joules_disk()
    dram = em.get_total_joules_dram()
    gpu = em.get_total_joules_gpu()

    total = cpu + disk + dram + gpu

    cpu_energy.append(cpu)
    disk_energy.append(disk)
    dram_energy.append(dram)
    gpu_energy.append(gpu)
    total_energy.append(total)

    print(f"CPU:   {cpu} J")
    print(f"Disk:  {disk} J")
    print(f"DRAM:  {dram} J")
    print(f"GPU:   {gpu} J")
    print(f"Total: {total} J")

# Calculate averages
avg_cpu = sum(cpu_energy) / num_iterations
avg_disk = sum(disk_energy) / num_iterations
avg_dram = sum(dram_energy) / num_iterations
avg_gpu = sum(gpu_energy) / num_iterations
avg_total = sum(total_energy) / num_iterations

print("\n==============================")
print("       AVERAGE (10 RUNS)")
print("==============================")

print(f"Average CPU:   {avg_cpu} J")
print(f"Average Disk:  {avg_disk} J")
print(f"Average DRAM:  {avg_dram} J")
print(f"Average GPU:   {avg_gpu} J")
print(f"Average Total: {avg_total} J")
```
